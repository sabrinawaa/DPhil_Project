```python

# In[2]:


## Bunch parameters
mass = RFT.electronmass # MeV/c^2
charge = -1 # single-particle charge, in units of e
population = 1e10 # number of real particles per bunch
Pc = 1000 # reference momentum, MeV/c
P_over_q = Pc / charge # MV/c, reference rigidity
twiss = RFT.Bunch6d_twiss()
twiss.emitt_x = 0.0001 # mm.mrad, normalized emittances
twiss.emitt_y = 0.0001 # mm.mrad
twiss.alpha_x = 0.0
twiss.alpha_y = 0.0
twiss.beta_x = 1
twiss.beta_y = 1
Npart = 5000 # number of macroparticles in the bunch
b0 = RFT.Bunch6d(mass, population, charge, Pc, twiss, Npart)
```
